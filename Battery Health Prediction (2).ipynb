### Data Preprocessing

```python
from sklearn.preprocessing import MinMaxScaler 
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['Remaining_Useful_Life'])
y = df['Remaining_Useful_Life']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=23)
```

```python
cols = ['Battery_Cycle_Count', 'Battery_Discharge_Duration_sec','Voltage_Drop_Time_sec', 'Maximum_Discharge_Voltage_V','Minimum_Charging_Voltage_V', 'Time_Near_Full_Charge_sec','Constant_Current_Charging_Time_sec', 'Total_Charging_Time_sec']
```

```python
num_pipeline = Pipeline([
    ("Imputation", SimpleImputer(strategy="median")),
    ("Scaling", MinMaxScaler())
])
```

```python
preprocessor = ColumnTransformer([("Numerical", num_pipeline, cols)])
preprocessor
```

```python
X_train=preprocessor.fit_transform(X_train)
X_test=preprocessor.transform(X_test)
```

### Simple Architecture

```python
from keras.models import Sequential
from keras.layers import Input,Dense
```

```python
model = Sequential([
    Input(shape=(X_train.shape[1],)),
    Dense(32,activation='relu'),
    Dense(1,activation='linear')
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
```

```python
model.fit(X_train,y_train,epochs=20,validation_data=(X_test,y_test),batch_size=264)
```

```python
import optuna
from keras.layers import Dropout,BatchNormalization
from keras.regularizers import l1_l2
from keras.optimizers import Adam

def objective(trail):
    lr_rate = trail.suggest_float('learning rate',1.e-4,1.e-2,log=True)
    n_layers = trail.suggest_int('n layers',1,10)
    epochs = trail.suggest_int('epochs',20,100)
    optimizers = trail.suggest_categorical('optimizer',['Adam','RMSprop','SGD'])
    activation = trail.suggest_categorical('activation',['relu'])

    model = Sequential()
    model.add(Input(shape=(X_train.shape[1],)))

    for i in range(n_layers):
        units = trail.suggest_int(f'units{i}',16,128)
        dropout = trail.suggest_float(f'dropout{i}',0,0.5)
        regu = trail.suggest_float('regularization',1.e-4,1.e-2)
        model.add(Dense(units,activation=activation,kernel_regularizer=l1_l2(l1=regu,l2=regu)))
        model.add(Dropout(dropout))
        model.add(BatchNormalization())

    model.add(Dense(1,activation='linear'))

    model.compile(optimizer=Adam(learning_rate = lr_rate),loss='mse',metrics=['mae'])

    model.fit(X_train,y_train,epochs=epochs,batch_size = 234,validation_data=(X_test,y_test))

    return min(model.history.history['val_mae'])
```

```python
study = optuna.create_study(direction='minimize')
study.optimize(objective,n_trials=10)
```

```python
study.best_params
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.regularizers import l1_l2
from tensorflow.keras.optimizers import Adam,RMSprop

model = Sequential()
model.add(Input(shape=(X_train.shape[1],)))

model.add(Dense(112,activation='relu',kernel_regularizer=l1_l2(l1=0.0038, l2=0.0038)))
model.add(BatchNormalization())
#model.add(Dropout(0.042))

model.add(Dense(98,activation='relu',kernel_regularizer=l1_l2(l1=0.0038, l2=0.0038)))
model.add(BatchNormalization())
#model.add(Dropout(0.42))

model.add(Dense(1, activation='linear'))
```

```python
model.summary()
```

```python
optimizer = RMSprop(learning_rate=0.0045)

model.compile(optimizer=optimizer,loss='mse',metrics=['mae','r2_score'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model.fit(X_train,y_train,epochs=81, batch_size=234,validation_data=(X_test, y_test),callbacks = [EarlyStopping(monitor = 'val_r2_score',mode = 'max',patience=5,min_delta=0.005,restore_best_weights=True)])
```
